```python
#7_ Normalize the following array to the range [0, 1]:

data = np.array([150, 220, 90, 300, 180, 45])
normalized_data = (data - data.min()) / (data.max() - data.min())
print(normalized_data)

```

```python
#8_  Standardize (z-score) the same array from question 7:

data = np.array([150, 220, 90, 300, 180, 45])
mean = data.mean()
z_scores = (data-mean) / data.std()
print(z_scores)

```

```python
# 9_ Given the 2D array:
  # a. Compute the mean of each row
  # b. Compute the sum of each column
  # c. Find the maximum value in the entire matrix and its position (row, column)

mat = np.array([
    [4, 8, 1, 12],
    [7, 3, 9, 5],
     [11, 2, 6, 10],
    [15, 0, 14, 13]])

# a. Compute the mean of each row
row_mean = mat.mean(axis=1)
print("a = ", row_mean)

# b. Compute the sum of each column
col_sum = mat.sum(axis=0)
print("b = ", col_sum)

# c. Find the maximum value in the entire matrix and its position (row, column)
max_value = mat.max()
row, col = np.unravel_index(mat.argmax(), mat.shape)
print(f"Max value = {max_value}, Position = (row={row}, column={col})")
```
