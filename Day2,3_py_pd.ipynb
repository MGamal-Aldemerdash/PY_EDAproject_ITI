```python
#8_ Select only the columns: Name, Grade, Subject, City

df = df[['Name', 'Grade', 'Subject', 'City']]
print(df)
```

```python
#9_ Show all students who Passed and have Grade ≥ 85.

df = pd.read_csv('students.csv')
df_Exc_Std = df[(df['Passed'] == True) & (df['Grade'] >= 85)]
print(df_Exc_Std[['Name', 'Grade', 'City', 'Passed']])
```

```python
#10- Show students whose Name contains the letter "a" or "A" (case insensitive)

Req_students = df[df['Name'].str.contains('a', case=False, na=False)]
print(Req_students)
```

```python
#11- Show students in Cairo or Alexandria with Grade < 80.

students = df[
    (df["City"].isin(["Cairo", "Alexandria"])) &
    (df["Grade"] < 80)]

print(students)
```

```python
#12_ •Show students with Grade between 80 and 90 (inclusive).

students = df[(df['Grade'] >= 80) & (df['Grade'] <= 90)]
print(students)
```

```python
#13_ Using .loc, change Khaled's grade from 55 to 68

df.loc[df['Name'] == 'Khaled', 'Grade'] = 68
print(df[df['Name']=='Khaled'])
```

```python
#14_ Add a new column Is_Excellent that is True if Grade ≥ 90, False otherwise (vectorized)

df["Is_Excellent"] = df["Grade"] >= 90
print(df.head())
```

```python
#15_ Fill missing Grade values with the average Grade

avg_grade = df['Grade'].mean()
df['Grade'] = df['Grade'].fillna(avg_grade)
print(f"Average Grade: {df['Grade'].mean():.2f}")
df.head(10)
```

```python
#16_ Fill missing Score Date values with the most common date (mode)

mode_date = df['Score Date'].mode()[0]
df['Score Date'] = df['Score Date'].fillna(mode_date)
df.head(10)
```

```python
#17_ Add a new column Final_Grade = Grade + 5 (bonus points).

df['Final_Grade'] = df['Grade'] + 5
print(df[['Name', 'Grade', 'Final_Grade']].head(10))
```

```python
#18_ Create a column Grade_Category using pd.cut with bins

bins = [0, 69, 79, 89, 100]
labels = ['Fail', 'Average', 'Good', 'Excellent']
df['Grade_Category'] = pd.cut(df['Grade'], bins=bins, labels=labels, include_lowest=True)
print(df[['Name', 'Grade', 'Grade_Category']].head(10))
```

```python
#19_ Remove the column Student ID

df.drop('Student ID', axis=1, inplace=True)
df.head(10)
```

```python
#20_ •Keep only students who Passed

df = df[df['Passed'] == True].reset_index(drop=True)
df.head(10)
```

```python
#21_ Sort the DataFrame by Grade descending

df = df.sort_values('Grade', ascending=False).reset_index(drop=True)
df.head(10)
```

```python
#22_ Sort by City ascending, then inside each city sort by Grade descending

df = df.sort_values(['City', 'Grade'], ascending=[True, False]).reset_index(drop=True)
df.head(10)
```

```python
#23_ Add a column Rank that ranks students by Grade (1 = highest, handle ties with 'min')

df['Rank'] = df['Grade'].rank(ascending=False, method='min')
df.head(10)
```

```python
#24_ Show the top 4 students with the highest grades (use nlargest)

top_4 = df.nlargest(4, 'Grade')
print(top_4[['Name', 'Grade']])
```

```python
#25_ Show the 3 youngest students (use nsmallest on Age)

youngest_3 = df.nsmallest(3, 'Age')
print(youngest_3[['Name', 'Age']])
```

```python
#26_ Find the student(s) with the highest Grade and show their Name and Grade

max_grade = df['Grade'].max()
df.head()
```

```python
#27_ After filling missing grades, recalculate the average Grade per Subject

avg_grade_per_subject = df.groupby('Subject')['Grade'].mean()
print(avg_grade_per_subject)
```

```python
#28_ •Create a new column City_Avg that shows the average grade of each student's city (using transform)

df['City_Avg'] = df.groupby('City')['Grade'].transform('mean')
df.head(10)
```

```python
#29_ Show the subject with the highest average grade

highest_average_subject = avg_grade_per_subject.idxmax()
print(f"Subject with highest average grade: {highest_average_subject}")
```
